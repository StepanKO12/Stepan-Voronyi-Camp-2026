# Optional hometask: Glasses

Implement a solution to glasses dropping problem.

Given:

number of floors
number of glasses

Determine the minimum number of attempts required to find the critical floor (the highest floor from which a glass does not break), considering both outcomes of each drop:

glass breaks

glass does not break

## Answer
The essence of the formula is to find the number of attempts in which we can check the maximum number of floors, given that we have a limited number of glasses.

```python
def min_attempts(k: int, F: int) -> int:
    dp = [0] * (k+1)
    m = 0
    while dp[k] < F:
        m += 1
        for i in range(k, 0, -1):
            dp[i] = dp[i] + dp[i-1] + 1
    return m

print(min_attempts(k=2, F=100))
print(min_attempts(k=3, F=100))
print(min_attempts(k=4, F=100))
print(min_attempts(k=3, F=1000))
print(min_attempts(k=5, F=1000))
```
